```python
import os, json, glob
import pandas as pd

DATA_DIR = "../analysis-of-corporate-reports-dataset"   # note the ../
ESG_ITEMS = ["item_1", "item_1A", "item_1C", "item_7"]

paths = glob.glob(DATA_DIR + "/*.json")
print(f"Starting to read {len(paths)} files...")

rows = []
for i, path in enumerate(paths):
    if i % 100 == 0:
        print(f"  ...read {i} files so far")
    try:
        with open(path) as f:
            rec = json.load(f)
        rows.append({
            "company": rec.get("company"),
            "year":    str(rec.get("period_of_report", ""))[:4],
            "sic":     rec.get("sic"),
            **{it: rec.get(it, "") or "" for it in ESG_ITEMS},
        })
    except Exception as e:
        print("skipped", os.path.basename(path), e)

df = pd.DataFrame(rows)
print(f"STEP 1 done: loaded {len(df)} filings, {df['company'].nunique()} companies")
```

```python
para_rows = []
for _, r in df.iterrows():
    for item in ESG_ITEMS:
        text = r[item] or ""
        for chunk in text.split(". "):
            chunk = chunk.strip()
            if len(chunk) > 100:               # drop tiny fragments
                para_rows.append({
                    "company": r["company"], "year": r["year"],
                    "item": item, "text": chunk
                })
paras = pd.DataFrame(para_rows)
print(f"STEP 2 done: {len(paras)} paragraphs")
```

```python
KEYWORDS = {
    "E": ["climate", "emission", "carbon", "net-zero", "net zero", "renewable",
          "greenhouse", "ghg", "pollution", "sustainab", "environment",
          "decarboniz", "clean energy", "biodiversity", "water", "waste",
          "energy efficien", "low-carbon", "deforestation"],
    "S": ["employee", "diversity", "inclusion", "equity", "community",
          "human capital", "human rights", "safety", "wellbeing", "well-being",
          "workforce", "labor", "labour", "philanthrop", "supply chain",
          "health and safety", "gender", "talent"],
    "G": ["board", "oversight", "ethics", "governance", "compliance",
          "accountab", "shareholder", "fiduciary", "audit committee",
          "anti-corruption", "bribery", "whistleblow", "transparency",
          "executive compensation", "proxy"],

}

def tag(text):
    t = text.lower()
    return [cat for cat, words in KEYWORDS.items() if any(w in t for w in words)]

paras["esg"] = paras["text"].apply(tag)
esg = paras[paras["esg"].map(len) > 0].copy()
print(f"STEP 3 done: {len(esg)} ESG paragraphs "
      f"({100*len(esg)//max(len(paras),1)}% of all paragraphs)")

esg.head(10)
```

```python
import os, json
import pandas as pd
from transformers import BertTokenizer, BertForSequenceClassification, pipeline
import torch

# ---- load the ESG model once ----
MODEL = "yiyanghkust/finbert-esg"
tokenizer = BertTokenizer.from_pretrained(MODEL)
model = BertForSequenceClassification.from_pretrained(MODEL, num_labels=4)
device = 0 if torch.cuda.is_available() else -1   # use GPU if present
esg_clf = pipeline("text-classification", model=model, tokenizer=tokenizer,
                   device=device, truncation=True, max_length=256)
print("Model loaded. GPU in use:", device == 0)

# ---- output folder for tagged files ----
OUT_DIR = "tagged_esg"
os.makedirs(OUT_DIR, exist_ok=True)

def tag_company(company_name, max_paras=None):
    """Tag all paragraphs for one company and save to its own JSON file."""
    sub = paras[paras["company"] == company_name].copy()
    if max_paras:                       # optional cap for speed while testing
        sub = sub.head(max_paras)
    if len(sub) == 0:
        print("No paragraphs for", company_name); return

    texts = sub["text"].tolist()
    preds = esg_clf(texts, batch_size=16)        # the model does the work
    sub["esg_label"] = [p["label"] for p in preds]
    sub["esg_score"] = [round(p["score"], 3) for p in preds]

    # keep only real ESG paragraphs (drop "None")
    sub_esg = sub[sub["esg_label"].str.lower() != "none"]

    safe = "".join(c if c.isalnum() else "_" for c in company_name)[:40]
    path = os.path.join(OUT_DIR, f"{safe}.json")
    sub_esg.to_json(path, orient="records", indent=2, force_ascii=False)
    print(f"{company_name}: {len(sub_esg)} ESG paragraphs saved -> {path}")
    return sub_esg

# ---- run it on ONE company first (fast, proves it works) ----c
result = tag_company("BlackRock, Inc.")   # or "WALT DISNEY CO/" etc.
result.head(10)
```

```python
def tag_company(company_name, max_paras=None, min_score=0.85):
    sub = paras[paras["company"] == company_name].copy()
    if max_paras:
        sub = sub.head(max_paras)
    if len(sub) == 0:
        print("No paragraphs for", company_name); return

    preds = esg_clf(sub["text"].tolist(), batch_size=16)
    sub["esg_label"] = [p["label"] for p in preds]
    sub["esg_score"] = [round(p["score"], 3) for p in preds]

    # keep only confident, non-None predictions
    keep = (sub["esg_label"].str.lower() != "none") & (sub["esg_score"] >= min_score)
    sub_esg = sub[keep]

    safe = "".join(c if c.isalnum() else "_" for c in company_name)[:40]
    path = os.path.join(OUT_DIR, f"{safe}.json")
    sub_esg.to_json(path, orient="records", indent=2, force_ascii=False)
    print(f"{company_name}: {len(sub_esg)} confident ESG paragraphs "
          f"(of {len(sub)} total) saved -> {path}")
    return sub_esg

result = tag_company("BlackRock, Inc.")
result.head(15)
```

```python
# drop the old keyword column — FinBERT's esg_label replaces it
if "esg" in paras.columns:
    paras = paras.drop(columns=["esg"])
```

```python
companies = paras["company"].unique()
print(f"Tagging {len(companies)} companies...")
for i, c in enumerate(companies):
    print(f"[{i+1}/{len(companies)}]", end=" ")
    tag_company(c)
```

```python
import os, json, glob
import pandas as pd

DATA_DIR = "../analysis-of-corporate-reports-dataset"   # the ../ path that worked
ESG_ITEMS = ["item_1", "item_1A", "item_1C", "item_7"]

paths = glob.glob(DATA_DIR + "/*.json")
rows = []
for path in paths:
    try:
        with open(path) as f:
            rec = json.load(f)
        rows.append({
            "company": rec.get("company"),
            "year":    str(rec.get("period_of_report", ""))[:4],
            "sic":     rec.get("sic"),
            **{it: rec.get(it, "") or "" for it in ESG_ITEMS},
        })
    except Exception as e:
        print("skipped", os.path.basename(path), e)
df = pd.DataFrame(rows)
print(f"Loaded {len(df)} filings, {df['company'].nunique()} companies")
```

```python
para_rows = []
for _, r in df.iterrows():
    for item in ESG_ITEMS:
        text = r[item] or ""
        for chunk in text.split(". "):
            chunk = chunk.strip()
            if len(chunk) > 100:
                para_rows.append({
                    "company": r["company"], "year": r["year"],
                    "item": item, "text": chunk
                })
paras = pd.DataFrame(para_rows)
print(f"Rebuilt paras: {len(paras)} paragraphs")
```

```python
import glob, pandas as pd

# 1. ESG paragraph counts per company-year (from tagged files)
frames = [pd.read_json(f) for f in glob.glob("tagged_esg/*.json")]
allesg = pd.concat(frames, ignore_index=True)
allesg = allesg[allesg["year"].astype(str).str.match(r"^20\d\d$")]
allesg["year"] = allesg["year"].astype(int)
esg_counts = allesg.groupby(["company", "year"]).size().rename("esg_paras")

# 2. TOTAL paragraph counts per company-year (from the full paras table)
paras_clean = paras[paras["year"].astype(str).str.match(r"^20\d\d$")].copy()
paras_clean["year"] = paras_clean["year"].astype(int)
total_counts = paras_clean.groupby(["company", "year"]).size().rename("total_paras")

# 3. combine and compute the share
share = pd.concat([esg_counts, total_counts], axis=1).fillna(0)
share["esg_share"] = share["esg_paras"] / share["total_paras"]
share = share.reset_index()

# 4. average ESG share across all companies, per year
yearly = share.groupby("year")["esg_share"].mean()

import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
(yearly*100).plot(marker="o")
plt.title("Average share of ESG language in S&P 100 filings over time")
plt.xlabel("Year"); plt.ylabel("ESG paragraphs as % of all paragraphs")
plt.grid(alpha=0.3)
plt.show()
```

```python
import glob, pandas as pd
import matplotlib.pyplot as plt

# load all tagged ESG paragraphs
frames = [pd.read_json(f) for f in glob.glob("tagged_esg/*.json")]
allesg = pd.concat(frames, ignore_index=True)
allesg = allesg[allesg["year"].astype(str).str.match(r"^20\d\d$")]
allesg["year"] = allesg["year"].astype(int)

# total paragraphs per year (denominator, from the full paras table)
paras_clean = paras[paras["year"].astype(str).str.match(r"^20\d\d$")].copy()
paras_clean["year"] = paras_clean["year"].astype(int)
totals = paras_clean.groupby("year").size()

# count E / S / G paragraphs per year, then turn into % of all paragraphs
by_cat = allesg.groupby(["year", "esg_label"]).size().unstack(fill_value=0)
by_cat_pct = by_cat.div(totals, axis=0) * 100      # normalize by total paragraphs

print(by_cat_pct.round(2))

# grouped bar chart: one cluster of E/S/G bars per year
by_cat_pct.plot(kind="bar", figsize=(12,6),
                color={"Environmental":"#2e7d32",
                       "Social":"#1565c0",
                       "Governance":"#f9a825"})
plt.title("Environmental / Social / Governance language over time (S&P 100)")
plt.xlabel("Year"); plt.ylabel("% of all paragraphs")
plt.legend(title="ESG category")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
import glob, pandas as pd
import matplotlib.pyplot as plt

# --- load tagged ESG paragraphs ---
frames = [pd.read_json(f) for f in glob.glob("tagged_esg/*.json")]
allesg = pd.concat(frames, ignore_index=True)
allesg = allesg[allesg["year"].astype(str).str.match(r"^20\d\d$")]
allesg["year"] = allesg["year"].astype(int)

# --- totals per year (denominator) ---
paras_clean = paras[paras["year"].astype(str).str.match(r"^20\d\d$")].copy()
paras_clean["year"] = paras_clean["year"].astype(int)
totals = paras_clean.groupby("year").size()

# --- E/S/G as % of all paragraphs, per year ---
by_cat = allesg.groupby(["year", "esg_label"]).size().unstack(fill_value=0)
by_cat_pct = by_cat.div(totals, axis=0) * 100

# overall ESG share = sum of the three pillars each year
overall_pct = by_cat_pct.sum(axis=1)

# --- plot: stacked bars + overall line ---
fig, ax = plt.subplots(figsize=(12,6))
by_cat_pct.plot(kind="bar", stacked=True, ax=ax,
                color={"Environmental":"#2e7d32",
                       "Social":"#1565c0",
                       "Governance":"#f9a825"}, alpha=0.85)

# overlay the overall average as a line (x positions 0,1,2... match the bars)
ax.plot(range(len(overall_pct)), overall_pct.values,
        color="black", marker="o", linewidth=2, label="Overall ESG share")

ax.set_title("ESG language over time: overall trend and E/S/G breakdown (S&P 100)")
ax.set_xlabel("Year"); ax.set_ylabel("% of all paragraphs")
ax.legend(title="")
ax.grid(axis="y", alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt

selected = {
    3: "Climate / emissions",
    2: "Diversity & inclusion",
    1: "COVID / pandemic",
    0: "Cybersecurity / governance",
    4: "Wellbeing / benefits",
}

fig, ax = plt.subplots(figsize=(11,6))
for topic_id, label in selected.items():
    sub = topics_over_time[topics_over_time["Topic"] == topic_id].sort_values("Timestamp")
    ax.plot(sub["Timestamp"], sub["Frequency"], marker="o", label=label, linewidth=2)

# event markers — vertical dashed lines with labels at the top
events = {2020: "COVID", 2022: "Anti-ESG backlash", 2023: "SEC cyber rule"}
ymax = ax.get_ylim()[1]                      # use the real top of the chart
for yr, label in events.items():
    ax.axvline(x=yr, color="gray", linestyle="--", alpha=0.6)
    ax.annotate(label, xy=(yr, ymax), xytext=(0, -5),
                textcoords="offset points",
                rotation=90, va="top", ha="right",
                fontsize=9, color="dimgray")

ax.set_title("How ESG themes in S&P 100 filings shifted over time")
ax.set_xlabel("Year"); ax.set_ylabel("Number of paragraphs")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```
