**Research question:** How does S&P 100 firms' sustainability discourse change across the decade, and do those changes coincide with external factors like the anti-ESG backlash, new regulations, or global events?

**Pipeline:** Extract → Tag (FinBERT-ESG) → Discover themes (BERTopic) → Track over time.

## Step 1 — Load all filings into one table
Read every JSON filing and keep the company, year, and the sections where sustainability content lives.

```python
import os, json, glob
import pandas as pd

DATA_DIR = "../analysis-of-corporate-reports-dataset"   # adjust if your path differs
ESG_ITEMS = ["item_1", "item_1A", "item_1C", "item_7"]  # sections that hold ESG text

paths = glob.glob(DATA_DIR + "/*.json")
print(f"Reading {len(paths)} files...")

rows = []
for i, path in enumerate(paths):
    if i % 200 == 0:
        print(f"  ...{i} files")
    try:
        with open(path) as f:
            rec = json.load(f)
        rows.append({
            "company": rec.get("company"),
            "year":    str(rec.get("period_of_report", ""))[:4],   # YYYY
            "sic":     rec.get("sic"),
            **{it: rec.get(it, "") or "" for it in ESG_ITEMS},
        })
    except Exception as e:
        print("skipped", os.path.basename(path), e)

df = pd.DataFrame(rows)
print(f"STEP 1 done: {len(df)} filings, {df['company'].nunique()} companies")
```

## Step 2 — Split filings into paragraphs
Each paragraph carries its company and year. Paragraphs are the unit of analysis.

```python
para_rows = []
for _, r in df.iterrows():
    for item in ESG_ITEMS:
        text = r[item] or ""
        for chunk in text.split(". "):
            chunk = chunk.strip()
            if len(chunk) > 100:                       # drop tiny fragments
                para_rows.append({"company": r["company"], "year": r["year"],
                                  "item": item, "text": chunk})
paras = pd.DataFrame(para_rows)
print(f"STEP 2 done: {len(paras)} paragraphs")
```

```python
df.to_parquet("filings_df.parquet")
paras.to_parquet("paras.parquet")
print("Saved — future runs can skip the slow read.")
```

## Step 3 — Tag each paragraph with FinBERT-ESG
A model trained on financial text labels each paragraph Environmental / Social / Governance / None.
We keep only confident (score >= 0.85), non-None predictions, and save one JSON file per company.

```python
from transformers import BertTokenizer, BertForSequenceClassification, pipeline
import torch

MODEL = "yiyanghkust/finbert-esg"
tokenizer = BertTokenizer.from_pretrained(MODEL)
model = BertForSequenceClassification.from_pretrained(MODEL, num_labels=4)
device = 0 if torch.cuda.is_available() else -1
esg_clf = pipeline("text-classification", model=model, tokenizer=tokenizer,
                   device=device, truncation=True, max_length=256)
print("Model loaded. GPU in use:", device == 0)

OUT_DIR = "tagged_esg"
os.makedirs(OUT_DIR, exist_ok=True)

def tag_company(company_name, min_score=0.85):
    """Tag one company's paragraphs and save the confident ESG ones to JSON."""
    sub = paras[paras["company"] == company_name].copy()
    if len(sub) == 0:
        print("No paragraphs for", company_name); return
    preds = esg_clf(sub["text"].tolist(), batch_size=16)
    sub["esg_label"] = [p["label"] for p in preds]
    sub["esg_score"] = [round(p["score"], 3) for p in preds]
    keep = (sub["esg_label"].str.lower() != "none") & (sub["esg_score"] >= min_score)
    sub_esg = sub[keep]
    safe = "".join(c if c.isalnum() else "_" for c in company_name)[:40]
    path = os.path.join(OUT_DIR, f"{safe}.json")
    sub_esg.to_json(path, orient="records", indent=2, force_ascii=False)
    print(f"{company_name}: {len(sub_esg)} confident ESG paragraphs (of {len(sub)}) -> {path}")
    return sub_esg

# quick test on one company to confirm it works
result = tag_company("BlackRock, Inc.")
result.head(10)
```

## Step 3b — Tag all companies
Runs the same function across every company. Saves per-company files so a crash never loses everything
(re-running skips companies already done).

```python
def already_done(name):
    safe = "".join(c if c.isalnum() else "_" for c in name)[:40]
    return os.path.exists(os.path.join(OUT_DIR, f"{safe}.json"))

companies = paras["company"].unique()
todo = [c for c in companies if not already_done(c)]
print(f"{len(companies)} companies, {len(companies)-len(todo)} done, {len(todo)} to go")
for i, c in enumerate(todo):
    print(f"[{i+1}/{len(todo)}]", end=" ")
    tag_company(c)
```

## Step 4 — Combine tagged files
Load every per-company file back into one table for analysis.

```python
frames = [pd.read_json(f) for f in glob.glob("tagged_esg/*.json")]
allesg = pd.concat(frames, ignore_index=True)
allesg = allesg[allesg["year"].astype(str).str.match(r"^20\d\d$")]
allesg["year"] = allesg["year"].astype(int)
print(f"Combined: {len(allesg)} ESG paragraphs across {allesg['company'].nunique()} companies")
```

## Chart 1 — Overall trend + E/S/G breakdown (normalized)
ESG language as a share of all paragraphs each year, split by category, with the overall trend on top.

```python
import matplotlib.pyplot as plt

# denominator: total paragraphs per year
paras_clean = paras[paras["year"].astype(str).str.match(r"^20\d\d$")].copy()
paras_clean["year"] = paras_clean["year"].astype(int)
totals = paras_clean.groupby("year").size()

# E/S/G as % of all paragraphs
by_cat = allesg.groupby(["year", "esg_label"]).size().unstack(fill_value=0)
by_cat_pct = by_cat.div(totals, axis=0) * 100
overall_pct = by_cat_pct.sum(axis=1)

fig, ax = plt.subplots(figsize=(12,6))
by_cat_pct.plot(kind="bar", stacked=True, ax=ax,
                color={"Environmental":"#2e7d32","Social":"#1565c0","Governance":"#f9a825"},
                alpha=0.85)
ax.plot(range(len(overall_pct)), overall_pct.values,
        color="black", marker="o", linewidth=2, label="Overall ESG share")
ax.set_title("ESG language over time: overall trend and E/S/G breakdown (S&P 100)")
ax.set_xlabel("Year"); ax.set_ylabel("% of all paragraphs")
ax.legend(title=""); ax.grid(axis="y", alpha=0.3)
plt.savefig("chart_esg_trend.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Step 5 — Discover themes with BERTopic
Automatically groups the ESG paragraphs into topics. Review the topic table, then note which topic
numbers correspond to the themes you care about (climate, diversity, COVID, cybersecurity, wellbeing).

```python
from bertopic import BERTopic

docs  = allesg["text"].tolist()
years = allesg["year"].tolist()

topic_model = BERTopic(min_topic_size=50, verbose=True)
topics, probs = topic_model.fit_transform(docs)
topic_model.get_topic_info().head(20)
```

## Step 6 — Themes over time
Compute each topic's prevalence per year. (This must run before any chart that uses `topics_over_time`.)

```python
topics_over_time = topic_model.topics_over_time(docs, years)

# interactive overview (saved to HTML so it always opens)
fig = topic_model.visualize_topics_over_time(topics_over_time, top_n_topics=10)
fig.write_html("themes_over_time.html")
print("Saved themes_over_time.html")
```

## Chart 2 — Headline: key ESG themes over time + event markers
Hand-pick the topic numbers that matter from the table above, give them readable names,
and overlay external events. **Update the `selected` numbers to match your Step 5 table.**

```python
import matplotlib.pyplot as plt

# EDIT these IDs to match your get_topic_info() table:
selected = {
    3: "Climate / emissions",
    2: "Diversity & inclusion",
    1: "COVID / pandemic",
    0: "Cybersecurity / governance",
    4: "Wellbeing / benefits",
}

fig, ax = plt.subplots(figsize=(11,6))
for topic_id, label in selected.items():
    sub = topics_over_time[topics_over_time["Topic"] == topic_id].sort_values("Timestamp")
    ax.plot(sub["Timestamp"], sub["Frequency"], marker="o", label=label, linewidth=2)

# external-event markers
events = {2020: "COVID", 2022: "Anti-ESG backlash", 2023: "SEC cyber rule"}
for yr in events:
    ax.axvline(x=yr, color="gray", linestyle="--", alpha=0.6)

ax.set_title("How ESG themes in S&P 100 filings shifted over time")
ax.set_xlabel("Year"); ax.set_ylabel("Number of paragraphs")
ax.legend(); ax.grid(alpha=0.3)
plt.savefig("chart_themes.png", dpi=150, bbox_inches="tight")
plt.show()
```

---
### Summary of findings
- **Finding 1:** ESG language roughly doubled (2016–2023), then cooled — Social dominates.
- **Finding 2:** Shifts track events — COVID spike (2020), cybersecurity surge (2022, SEC rule),
  and climate/diversity cooling after the 2022 anti-ESG backlash.
- **Insight:** Sustainability disclosure is strategic communication that reacts to the world,
  not a steady mirror of values.

**Limitations:** we measure words, not deeds (not proof of greenwashing); "coincide" is not "cause";
FinBERT leans toward Social.
